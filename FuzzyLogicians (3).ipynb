```python
import os
import io
import cv2
import re
import torch
import threading
import nest_asyncio
import uvicorn
import numpy as np

from fastapi import FastAPI, File, UploadFile, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pyzbar.pyzbar import decode
from PIL import Image
from qwen_vl_utils import process_vision_info


# ============================================================
# CONFIGURATION
# ============================================================

HOST = "0.0.0.0"
PORT = 8000

MODEL_LOCK = threading.Lock()


# ============================================================
# FASTAPI
# ============================================================

nest_asyncio.apply()

app = FastAPI(
    title="Legal Metrology Compliance Scanner",
    description="AI-powered product label compliance scanner",
    version="1.0.0"
)

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"]
)


# ============================================================
# PROMPT - UPDATED FOR STRICT NUMERIC MRP
# ============================================================

OCR_PROMPT = """
You are an OCR and Legal Metrology compliance assistant.
Read the product packaging image carefully.

STEP 1: Transcription
Transcribe all readable text from the image.

STEP 2: Extraction
Extract the 6 mandatory declarations. If a detail is missing, blank, or unreadable on the package, output exactly MISSING.
For the MRP, you MUST extract the numerical price. If there are no numbers printed next to the MRP label, output MISSING.

You MUST wrap your extracted values exactly in the following XML tags:

<MANUFACTURER> value here </MANUFACTURER>
<COMMODITY> value here </COMMODITY>
<NET_QUANTITY> value here </NET_QUANTITY>
<DATE> value here </DATE>
<MRP> value here </MRP>
<CONSUMER_CARE> value here </CONSUMER_CARE>
"""


# ============================================================
# BARCODE SCANNER
# ============================================================

def scan_barcode_bytes(image_bytes):

    try:
        image_array = cv2.imdecode(np.frombuffer(image_bytes, dtype=np.uint8), cv2.IMREAD_COLOR)

        if image_array is None:
            return []

        gray = cv2.cvtColor(image_array, cv2.COLOR_BGR2GRAY)
        blurred = cv2.GaussianBlur(gray, (3, 3), 0)
        otsu = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
        adaptive = cv2.adaptiveThreshold(blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 31, 5)

        variants = [image_array, gray, otsu, adaptive]
        detected = []

        for variant in variants:
            try:
                codes = decode(variant)
                for code in codes:
                    try:
                        data = code.data.decode("utf-8", errors="ignore")
                    except Exception:
                        data = str(code.data)

                    item = {"type": code.type, "data": data}
                    if item not in detected:
                        detected.append(item)
            except Exception:
                continue

        return detected

    except Exception as error:
        print(f"[Barcode] Error: {error}")
        return []


# ============================================================
# MISSING VALUE HANDLING
# ============================================================

MISSING_TOKENS = {
    "missing", "not found", "none", "null", "n/a", "na", "", "[missing]", "[not found]"
}


def normalize_value(value):
    if value is None:
        return None
    value = str(value).strip()
    if value.lower() in MISSING_TOKENS:
        return None
    return value


def find_net_quantity_from_text(text):
    if not text:
        return None
    patterns = [
        r'\bNET\s*(?:QTY|QUANTITY)\s*[:\-]?\s*([0-9]+(?:\.[0-9]+)?\s*(?:kg|g|mg|l|ml|cl|ml|litre|liter|litres|liters|pcs|pc|pieces|piece|nos|no))',
        r'\bNET\s*QUANTITY\s*[:\-]?\s*([0-9]+(?:\.[0-9]+)?\s*(?:kg|g|mg|l|ml|cl|litre|liter|litres|liters|pcs|pc|pieces|piece|nos|no))',
        r'\bNET\s*(?:WT|WEIGHT)\s*[:\-]?\s*([0-9]+(?:\.[0-9]+)?\s*(?:kg|g|mg|l|ml|cl|litre|liter|litres|liters|pcs|pc|pieces|piece|nos|no))',
        r'\bNET\b[^\n]{0,30}?([0-9]+(?:\.[0-9]+)?\s*(?:kg|g|mg|l|ml|cl|litre|liter|litres|liters|pcs|pc|pieces|piece|nos|no))'
    ]
    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE)
        if match:
            return match.group(1).strip()
    return None


def find_mrp_from_text(text):
    if not text:
        return None
    patterns = [
        r'\bMRP\b[^\n₹Rs.]{0,20}(?:₹|Rs\.?|INR)\s*([0-9]+(?:\.[0-9]+)?)',
        r'\bMRP\b\s*[:\-]?\s*([0-9]+(?:\.[0-9]+)?)',
        r'\bMAXIMUM\s+RETAIL\s+PRICE\b\s*[:\-]?\s*(?:₹|Rs\.?|INR)?\s*([0-9]+(?:\.[0-9]+)?)'
    ]
    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE)
        if match:
            return match.group(1).strip()
    return None


# ============================================================
# PARSE QWEN RESPONSE - WITH NUMBER VALIDATION
# ============================================================

def parse_model_output(raw_output):
    fields = {
        "manufacturer_details": None,
        "commodity_name": None,
        "net_quantity": None,
        "date_of_manufacture": None,
        "retail_sale_price": None,
        "consumer_care": None
    }

    tags_map = {
        "manufacturer_details": "MANUFACTURER",
        "commodity_name": "COMMODITY",
        "net_quantity": "NET_QUANTITY",
        "date_of_manufacture": "DATE",
        "retail_sale_price": "MRP",
        "consumer_care": "CONSUMER_CARE"
    }

    # 1. Regex Extraction for the 6 specific rules
    for field_key, tag in tags_map.items():
        pattern = rf"<{tag}>(.*?)</{tag}>"
        match = re.search(pattern, raw_output, re.DOTALL | re.IGNORECASE)

        if match:
            val = match.group(1).strip()
            fields[field_key] = normalize_value(val)

    # 2. Isolate the detected text
    detected_text = re.sub(r"<[^>]+>.*?</[^>]+>", "", raw_output, flags=re.DOTALL | re.IGNORECASE)
    detected_text = detected_text.replace("STEP 1: Transcription", "").replace("STEP 2: Extraction", "").strip()

    # 3. STRICT VALIDATION: MRP must contain a number
    mrp_val = fields.get("retail_sale_price")
    if mrp_val is not None:
        # If there are no digits anywhere in the MRP string, reject it
        if not any(char.isdigit() for char in mrp_val):
            fields["retail_sale_price"] = None

    # 4. Deterministic Fallbacks
    if fields["net_quantity"] is None:
        recovered_net_qty = find_net_quantity_from_text(detected_text)
        if recovered_net_qty:
            fields["net_quantity"] = recovered_net_qty
            print(f"[RULE FALLBACK] Net quantity recovered: {recovered_net_qty}")

    if fields["retail_sale_price"] is None:
        recovered_mrp = find_mrp_from_text(detected_text)
        if recovered_mrp:
            fields["retail_sale_price"] = recovered_mrp
            print(f"[RULE FALLBACK] MRP recovered: {recovered_mrp}")

    return detected_text, fields


# ============================================================
# LEGAL METROLOGY RULES
# ============================================================

RULES_CHECK = {
    "manufacturer_details": "Rule 6(1)(a): Name & Address of Mfg/Packer",
    "commodity_name": "Rule 6(1)(b): Generic/Common Commodity Name",
    "net_quantity": "Rule 6(1)(c): Net Quantity (Weight/Measure/Count)",
    "date_of_manufacture": "Rule 6(1)(d): Month & Year of Packing/Mfg",
    "retail_sale_price": "Rule 6(1)(e): Maximum Retail Price (MRP)",
    "consumer_care": "Rule 6(2): Consumer Complaint Contact Details"
}


# ============================================================
# COMPLIANCE EVALUATION
# ============================================================

def evaluate_compliance(fields):
    passed = []
    violations = []

    for key, rule_label in RULES_CHECK.items():
        value = fields.get(key)
        if value is not None:
            passed.append({"field": key, "rule": rule_label, "value": value})
        else:
            violations.append({"field": key, "rule": rule_label, "value": None})

    compliant = len(violations) == 0

    return {
        "status": "COMPLIANT" if compliant else "NON-COMPLIANT",
        "passed": passed,
        "violations": violations,
        "passed_count": len(passed),
        "violation_count": len(violations),
        "total_checks": len(RULES_CHECK),
        "compliant": compliant,
        "overall_evaluation": "COMPLIANT: All mandatory declarations are present." if compliant else "NON-COMPLIANT: Missing or incomplete statutory declarations."
    }


# ============================================================
# QWEN VISION PROCESSING
# ============================================================

def extract_compliance(image_bytes):

    if model is None or processor is None:
        raise RuntimeError("Qwen model is not loaded.")

    try:
        image = Image.open(io.BytesIO(image_bytes)).convert("RGB")
    except Exception as error:
        raise ValueError(f"Invalid image: {error}")

    messages = [
        {
            "role": "user",
            "content": [
                {"type": "image", "image": image},
                {"type": "text", "text": OCR_PROMPT}
            ]
        }
    ]

    text_prompt = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    image_inputs, video_inputs = process_vision_info(messages)

    inputs = processor(
        text=[text_prompt],
        images=image_inputs,
        videos=video_inputs,
        padding=True,
        return_tensors="pt"
    )

    if torch.cuda.is_available():
        inputs = {
            key: value.to(model.device) if hasattr(value, "to") else value
            for key, value in inputs.items()
        }

    print("[Qwen] Generating OCR + compliance response...")

    with MODEL_LOCK:
        with torch.inference_mode():
            generated_ids = model.generate(
                **inputs,
                max_new_tokens=1024,
                do_sample=False,
                repetition_penalty=1.15
            )

    generated_ids_trimmed = [
        out_ids[len(in_ids):]
        for in_ids, out_ids in zip(inputs["input_ids"], generated_ids)
    ]

    raw_output = processor.batch_decode(generated_ids_trimmed, skip_special_tokens=True)[0].strip()

    print("[Qwen] Response received.")

    detected_text, fields = parse_model_output(raw_output)
    compliance = evaluate_compliance(fields)

    return {
        "status": compliance["status"],
        "fields": fields,
        "detected_text": detected_text,
        "raw_model_output": raw_output,
        "passed": compliance["passed"],
        "violations": compliance["violations"],
        "passed_count": compliance["passed_count"],
        "violation_count": compliance["violation_count"],
        "total_checks": compliance["total_checks"],
        "compliant": compliance["compliant"],
        "overall_evaluation": compliance["overall_evaluation"]
    }


# ============================================================
# ROUTES
# ============================================================

@app.get("/")
def root():
    return {
        "status": "online",
        "service": "Legal Metrology Compliance Scanner",
        "model": "Qwen2.5-VL-7B-Instruct",
        "quantization": "4-bit NF4"
    }

@app.get("/health")
def health():
    gpu_name = None
    if torch.cuda.is_available():
        try:
            gpu_name = torch.cuda.get_device_name(0)
        except Exception:
            gpu_name = "CUDA GPU"

    return {
        "status": "healthy",
        "cuda": torch.cuda.is_available(),
        "gpu": gpu_name,
        "model_loaded": model is not None,
        "processor_loaded": processor is not None,
        "quantization": "4-bit NF4"
    }


@app.post("/scan")
async def scan_product(file: UploadFile = File(...)):
    if not file.content_type:
        raise HTTPException(status_code=400, detail="Invalid file.")

    if not file.content_type.startswith("image/"):
        raise HTTPException(status_code=400, detail="Please upload an image file.")

    image_bytes = await file.read()
    if not image_bytes:
        raise HTTPException(status_code=400, detail="Empty image.")

    print("\n" + "=" * 60)
    print(f"Scanning: {file.filename}")
    print(f"Image size: {len(image_bytes) / 1024:.1f} KB")
    print("=" * 60)

    print("[Barcode] Scanning...")
    barcodes = scan_barcode_bytes(image_bytes)
    print(f"[Barcode] {len(barcodes)} detected.")

    print("[Qwen] Running vision model...")
    try:
        compliance = extract_compliance(image_bytes)
    except ValueError as error:
        raise HTTPException(status_code=400, detail=str(error))
    except RuntimeError as error:
        raise HTTPException(status_code=503, detail=str(error))
    except torch.cuda.OutOfMemoryError:
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
        raise HTTPException(status_code=503, detail="GPU out of memory. Try a smaller image.")
    except Exception as error:
        print(f"[Qwen] ERROR: {error}")
        raise HTTPException(status_code=500, detail=f"AI processing failed: {error}")

    response = {
        "filename": file.filename,
        "barcode": {
            "detected": len(barcodes) > 0,
            "items": barcodes
        },
        "compliance": compliance
    }

    print(f"[Result] {compliance['status']}")
    print(f"[Result] Passed: {compliance['passed_count']}/{compliance['total_checks']}")
    print(f"[Result] Detected text characters: {len(compliance['detected_text'])}")
    print("=" * 60 + "\n")

    return response


# ============================================================
# APPLICATION READY
# ============================================================

print("=" * 60)
print("LEGAL METROLOGY COMPLIANCE SCANNER")
print("=" * 60)
print(f"CUDA available: {torch.cuda.is_available()}")

if torch.cuda.is_available():
    try:
        print(f"GPU: {torch.cuda.get_device_name(0)}")
    except Exception:
        pass

print("FastAPI application created successfully.")
print("Qwen model must already be loaded.")
print("=" * 60)
```
